```python
import pandas as pd
import os

movies = pd.read_csv("../data/movies_cleaned.csv")

print("Dataset Shape:", movies.shape)
print("\nDataset Columns:")
print(movies.columns.tolist())

display(movies.head())
```

```python
top_revenue = movies[
    movies["revenue"] > 0
].sort_values(
    by="revenue",
    ascending=False
).head(10)

display(top_revenue[["title", "budget", "revenue", "vote_average"]])
```

```python
top_popular = movies[
    movies["popularity"] > 0
].sort_values(
    by="popularity",
    ascending=False
).head(10)

display(
    top_popular[
        ["title", "popularity", "vote_average", "vote_count"]
    ]
)
```

```python
valid_movies = movies[
    (movies["budget"] > 0) &
    (movies["revenue"] > 0)
]

average_budget = valid_movies["budget"].mean()
average_revenue = valid_movies["revenue"].mean()

print("Average Budget:", round(average_budget, 2))
print("Average Revenue:", round(average_revenue, 2))
```

```python
movies["release_year"] = pd.to_datetime(
    movies["release_date"],
    errors="coerce"
).dt.year

movies_per_year = (
    movies.dropna(subset=["release_year"])
    .groupby("release_year")
    .size()
    .reset_index(name="movie_count")
)

movies_per_year["release_year"] = (
    movies_per_year["release_year"].astype(int)
)

display(movies_per_year)
```

```python
top_rated = movies[
    movies["vote_count"] >= 100
].sort_values(
    by=["vote_average", "vote_count"],
    ascending=[False, False]
).head(10)

display(
    top_rated[
        ["title", "vote_average", "vote_count", "popularity"]
    ]
)
```

```python
valid_movies = movies[
    (movies["budget"] > 0) &
    (movies["revenue"] > 0)
].copy()

valid_movies["ROI"] = (
    valid_movies["revenue"] / valid_movies["budget"]
).round(2)

top_roi = valid_movies.sort_values(
    by="ROI",
    ascending=False
).head(10)

display(
    top_roi[
        ["title", "budget", "revenue", "ROI"]
    ]
)
```

```python
language_revenue = (
    movies[movies["revenue"] > 0]
    .groupby("original_language")
    .agg(
        movie_count=("movie_id", "count"),
        average_revenue=("revenue", "mean"),
        total_revenue=("revenue", "sum")
    )
    .sort_values(
        by="total_revenue",
        ascending=False
    )
    .head(10)
    .reset_index()
)

language_revenue["average_revenue"] = (
    language_revenue["average_revenue"].round(2)
)

display(language_revenue)
```

```python
status_analysis = (
    movies.groupby("status")
    .agg(
        movie_count=("movie_id", "count"),
        average_rating=("vote_average", "mean"),
        total_votes=("vote_count", "sum")
    )
    .reset_index()
)

status_analysis["average_rating"] = (
    status_analysis["average_rating"].round(2)
)

display(status_analysis)
```

```python
longest_movies = (
    movies[movies["runtime"] > 0]
    .sort_values(
        by="runtime",
        ascending=False
    )
    .head(10)
)

display(
    longest_movies[
        ["title", "runtime", "release_date", "vote_average"]
    ]
)
```

```python
most_voted_movies = (
    movies[movies["vote_count"] > 0]
    .sort_values(
        by="vote_count",
        ascending=False
    )
    .head(10)
)

display(
    most_voted_movies[
        ["title", "vote_count", "vote_average", "popularity"]
    ]
)
```
