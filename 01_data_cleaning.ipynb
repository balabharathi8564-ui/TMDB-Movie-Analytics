```python
import pandas as pd
import os

data_folder = "../data"

movies = pd.read_csv(os.path.join(data_folder, "movies.csv"))

print("MOVIES DATASET")
print("Rows and Columns:", movies.shape)
print("\nColumn Names:")
print(movies.columns.tolist())

movies.head()
```

```python
# Check missing values in the movies dataset

print("MISSING VALUES")
print(movies.isnull().sum())

print("\nDUPLICATE MOVIE RECORDS")
print(movies.duplicated().sum())

print("\nDATASET INFORMATION")
movies.info()
```

```python
movies = movies.drop_duplicates()

print("Duplicates after cleaning:", movies.duplicated().sum())
print("Updated dataset shape:", movies.shape)
```

```python
# Check missing values after duplicate removal

print("Missing values in each column:")
print(movies.isnull().sum())

print("\nTotal missing values:")
print(movies.isnull().sum().sum())
```
